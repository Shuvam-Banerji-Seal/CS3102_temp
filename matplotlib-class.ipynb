### Different plots

We have `bar()`, `plot()`, `scatter()` and `pie()` that make it easy to create bar charts, line plots, scatter graphs and pie charts respectively.


The pyplot API can be used to
* construct the plot,
* configure labels and axis,
* manage color and line styles,
* handles events/allows plots to be interactive,
* display (show) the plot.

### Line Graph

```python
# X and Y coordinate of the data
x = [0, 1, 2, 3, 4, 5, 6]
y = [0, 2, 6, 14, 30, 43, 75]

# Set the axes labels
plt.ylabel('Speed', fontsize=20, color="cyan")
plt.xlabel('Time', fontsize=12)

# Set the title
plt.title("Speed vs Time")

# Plot and display the graph
# Using blue (b) circles (o) for markers ('bo')
# and a solid line ('-')
plt.plot(x, y, 'gv:')
plt.show()
```

```python
## another way of doing the same

# X and Y coordinate of the data
speed_time = [(0, 1, 2, 3, 4, 5, 6), (0, 2, 6, 14, 30, 43, 75)]

# Set the axes labels
plt.ylabel('Speed', fontsize=12, color="cyan")
plt.xlabel('Time', fontsize=12)

# Set the title
plt.title("Speed vs Time")

# Plot and display the graph
# Using blue (b) circles (o) for markers ('bo')
# and a solid line ('-')
plt.plot(speed_time[0], speed_time[1], 'ro-')
plt.show()
plt.scatter(speed_time[0], speed_time[1])
plt.show()
```

The plotting is done using the `pyplot.plot()` function. This function can take a wide range of parameters, the only **compulsory** parameters being the data used (here `X` and `Y`) to define the plot points. In the above example a third parameter is provided; this is a string `bo-`. This is a _coded format string_ in that each element of the string is meaningful to the `pyplot.plot()` function.

### fit a trend line in scatter plot

```python
x = (5, 5.5, 6, 6.5, 7, 8, 9, 10)
y = (120, 115, 100, 112, 80, 85, 69, 65)

# Generate the scatter plot
plt.scatter(x, y)

# Generate the trend line
z = np.polyfit(x, y, 5)
a = np.poly1d(z)
plt.plot(x, a(x), 'g')

# Display the figure
plt.show()
```

The `polyfit()` function performs a least squared polynomial fit for the data it is given. A `poly1d` class is then created based on the array returned by `polyfit()`. This class is a one-dimensional polynomial class. It is a convenience class, used to encapsulate natural operations on polynomials. The `poly1d` object is then used to generate a set of values for use with the set of x values for the function `pyplot.plot()`.

### Pie chart

```python
import matplotlib.pyplot as pyplot

labels = ('Python', 'Java', 'Scala', 'C#')
sizes = [45, 30, 15, 10]
pyplot.pie(sizes, labels=labels, autopct='%1.f%%', counterclock=True, startangle=130)
pyplot.show()

# only "explode" the 1st slice (i.e. 'Python')
explode = (0.1, 0, 0, 0)
pyplot.pie(sizes,
explode=explode,
labels=labels,
autopct='%1.f%%',
shadow=True,
counterclock=False,
startangle=90)
pyplot.show()
```

Parameters of `pyplot.pie()`:

* `labels`: an optional parameter that can take a sequence of strings that are used to provide labels for each wedge.
* `autopct`: takes a string (or function) to be used to format the numeric values used with each wedge.
* `counterclockwise`: By default wedges are plotted counter clockwise in pyplot and so to ensure that the layout is more like the traditional clockwise approach the `counterclock` parameter is set to `False`.
* `startangle`:. The starting angle has also been moved 90° using the `startangle` parameter so that the first segment starts at the top of the chart.
* `explode`: takes a sequence of values indicating how much a segment should be exploded by.

#### Stacked Bar Charts

```python
import matplotlib.pyplot as pyplot

# Set up the data
labels = ('Python', 'Scala', 'C#', 'Java', 'PHP')
index = (1, 2, 3, 4, 5)
web_usage = [20, 2, 5, 10, 14]
data_science_usage = [15, 8, 5, 15, 2]
games_usage = [10, 1, 5, 5, 4]

# Set up the bar chart
pyplot.bar(index, web_usage, tick_label=labels, label='web')
pyplot.bar(index, data_science_usage, tick_label=labels, label='data science', bottom=web_usage)
web_and_games_usage = [web_usage[i] + data_science_usage[i]
                       for i in range(0, len(web_usage))]
pyplot.bar(index, games_usage, tick_label=labels, label='games', bottom=web_and_games_usage)

# Configure the layout
pyplot.ylabel('Usage')
pyplot.xlabel('Programming Languages')
pyplot.legend()

# Display the chart
pyplot.show()
```

#### Grouped bar chart

```python
import matplotlib.pyplot as pyplot
BAR_WIDTH = 0.35
# set up grouped bar charts
teama_results = (60, 75, 56, 62, 58)
teamb_results = (55, 68, 80, 73, 55)
# Set up the index for each bar
index_teama = (1, 2, 3, 4, 5)
index_teamb = [i + BAR_WIDTH for i in index_teama]
# Determine the mid point for the ticks
ticks = [i + BAR_WIDTH / 2 for i in index_teama]
tick_labels = ('Lab 1', 'Lab 2', 'Lab 3', 'Lab 4', 'Lab 5')
# Plot the bar charts
pyplot.bar(index_teama, teama_results, BAR_WIDTH, color='b',
label='Team A')
pyplot.bar(index_teamb, teamb_results, BAR_WIDTH, color='g',
label='Team B')
# Set up the graph
pyplot.xlabel('Labs')
pyplot.ylabel('Scores')
pyplot.title('Scores by Lab')
pyplot.xticks(ticks, tick_labels)
pyplot.legend()
# Display the graph
pyplot.show()
```

## Annotate plots

There are two ways of doing this in matplotlib:

1. Using the `text()` function
2. Using the `annotate()` function of pyplot

### Using `text()`

It takes x and y position as arguments and lets you write text at coordinate `(x,y)`. It is particularly beneficial in case we want to mention values over each bar plot:

```python
fig=plt.figure(figsize=(6,6))

temp = []
for i in range(5):
    temp+= [random.uniform(1,20)]
# shortcut:
# temp=[ random.uniform(1,20) for i in range(5)]
# print(temp)

city=['City A','City B','City C','City D','City E']
y_pos=list(range(1,6))

graph=plt.bar(y_pos, temp, color='blue')

plt.xticks(y_pos,city)
plt.title('City Temperature')
plt.xlabel('Cities')
plt.ylabel('Temperature ($^\circ$C)')

for bar,t in zip(graph,temp):
    plt.text(bar.get_x() + bar.get_width()/2.0,bar.get_height(),'%.2f $^\circ$C'%t,ha='center',va='bottom')

plt.show()
```

If we want to annotate a point on the plot with a marker and also write text, we can use `annotate()` function like this:

```python
fig=plt.figure(figsize=(8,6))

X = list(range(1,11))
Y = []
Z = []
for x in X:
    Y += [x ** 2]

plt.plot(X, Y)
# plt.plot(X,np.exp(X))
plt.title('Annotating Exponential Plot using plt.annotate()')
plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.annotate('(3, 9)',xy=(3,9),arrowprops=dict(arrowstyle='->'),xytext=(2,15))

plt.annotate('(5, 25)',xy=(5,25),arrowprops=dict(arrowstyle='->',connectionstyle='arc3,rad=-.2'),xytext=(4,50))

plt.annotate('(8, 64)',xy=(8,64),arrowprops=dict(arrowstyle='-|>',connectionstyle='angle,angleA=90,angleB=0'),xytext=(5.5,70))

plt.show()
```

# ploting an expression

```python
import numpy as np  
import matplotlib.pyplot as plt  
from math import sin
def graph(formula, x_range):
    x = np.array(x_range)
    y = eval(formula)

    plt.plot(x, y)  
    plt.show()
```

```python
graph('x**3+2*x-4', range(-10, 11))
graph('x**2+(x+1)', range(-10, 11))
graph('np.sin(x)**2+np.sin(x+1)', np.linspace(0, 3*np.pi, 500))
```

```python
# implement the example graphs/integral from pyx
# from pylab import *
from matplotlib.patches import Polygon
import numpy as np
import matplotlib.pyplot as plt

def func(x):
    return x**3 - 15*x*x + 71*x - 20

ax = plt.subplot(111)

a, b = 2, 9 # integral area
x = np.arange(0, 10, 0.01)
y = func(x)
# plot(x, y, linewidth=1)

# make the shaded region
ix = np.arange(a, b, 0.01)
iy = func(ix)
verts = [(a,0)] + list(zip(ix,iy)) + [(b,0)]
poly = Polygon(verts, facecolor='0.8', edgecolor='k')
ax.add_patch(poly)

plt.text(0.5 * (a + b), 30,
     r"$\int_a^b f(x)\mathrm{d}x$", horizontalalignment='center',
     fontsize=20)

plt.axis([0,10, 0, 180])
plt.figtext(0.9, 0.05, 'x')
plt.figtext(0.1, 0.9, 'y')
# ax.set_xticks((a,b))
# ax.set_xticklabels(('a','b'))
# ax.set_yticks([])
ax.set_xlabel = 'sdf'
ax.set_ylabel = 'ad'
plt.show()
```
